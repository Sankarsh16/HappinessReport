```python
whr.loc[whr['Region'].isnull(), 'Region'] = 'Unknown'
```

```python
whr.rename(columns={'Explained by: Log GDP per capita': 'log gdp per capita','Explained by: Social support':'social support'}, inplace=True)
```

```python
whr.rename(columns={'Explained by: Healthy life expectancy': 'healthy life expectancy','Explained by: Freedom to make life choices':'freedom to make life choices'}, inplace=True)
```

```python
whr.rename(columns={'Explained by: Generosity': 'generosity','Explained by: Perceptions of corruption':'perceptions of corruption'}, inplace=True)
```

```python
whr.describe()
```

```python
whr['log gdp per capita'].fillna(whr['log gdp per capita'].mean(), inplace=True)
```

```python
whr['social support'].fillna(whr['social support'].mean(), inplace=True)
```

```python
whr['healthy life expectancy'].fillna(whr['healthy life expectancy'].mean(), inplace=True)
```

```python
whr['freedom to make life choices'].fillna(whr['freedom to make life choices'].mean(), inplace=True)
```

```python
whr['generosity'].fillna(whr['generosity'].mean(), inplace=True)
```

```python
whr['perceptions of corruption'].fillna(whr['perceptions of corruption'].mean(), inplace=True)
```

```python
whr['Dystopia + residual'].fillna(whr['Dystopia + residual'].mean(), inplace=True)
```

```python
whr['Total_explained'] = (
    whr['log gdp per capita'] +
    whr['social support'] +
    whr['healthy life expectancy'] +
    whr['freedom to make life choices'] +
    whr['generosity'] +
    whr['perceptions of corruption']
)
```

```python
whr['GDP_contribution_ratio'] = whr['log gdp per capita'] / whr['Total_explained']
whr['Social_support_ratio'] = whr['social support'] / whr['Total_explained']
whr['Healthy_life_expectancy_ratio'] = whr['healthy life expectancy'] / whr['Total_explained']
whr['Freedom_to_make_life_choices_ratio'] = whr['freedom to make life choices'] / whr['Total_explained']
whr['Generosity_ratio'] = whr['generosity'] / whr['Total_explained']
whr['Perceptions_of_corruption_ratio'] = whr['perceptions of corruption'] / whr['Total_explained']
```

```python
contribution_means = whr[['GDP_contribution_ratio', 'Social_support_ratio',
                          'Healthy_life_expectancy_ratio', 'Freedom_to_make_life_choices_ratio',
                          'Generosity_ratio', 'Perceptions_of_corruption_ratio']].mean()


plt.figure(figsize=(10, 6))
ax=contribution_means.plot(kind='bar', color=['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b'])
plt.title('Average Contribution Ratios of Happiness Factors')
plt.ylabel('Average Contribution Ratio')
plt.xlabel('Happiness Factors')
plt.xticks(rotation=90)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
for container in ax.containers:
    ax.bar_label(container)
plt.show()
```

```python
if 'Region' in whr.columns:
    regional_means = whr.groupby('Region')[['GDP_contribution_ratio', 'Social_support_ratio',
                                            'Healthy_life_expectancy_ratio', 'Freedom_to_make_life_choices_ratio',
                                            'Generosity_ratio', 'Perceptions_of_corruption_ratio']].mean()


    regional_means.plot(kind='bar', figsize=(14, 8), colormap='tab10')
    plt.title('Average Contribution Ratios by Region')
    plt.ylabel('Average Contribution Ratio')
    plt.xlabel('Region')
    plt.xticks(rotation=90)
    plt.legend(title='Factors', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
```

```python
import seaborn as sns

subset = whr.nlargest(10, 'Ladder score').set_index('Country')


plt.figure(figsize=(12, 8))
sns.heatmap(subset[['GDP_contribution_ratio', 'Social_support_ratio', 'Healthy_life_expectancy_ratio',
                    'Freedom_to_make_life_choices_ratio', 'Generosity_ratio', 'Perceptions_of_corruption_ratio']],
            annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Contribution Ratios for Top 10 Happiest Countries')
plt.ylabel('Country')
plt.xlabel('Happiness Factors')
plt.tight_layout()

```

```python
usa_values24=whr[whr['Country'] == 'United States']
```

```python
usa_values23['Healthy life expectancy']=usa_values23['Healthy life expectancy']/100


comparison_df = pd.DataFrame({
    '2023': usa_values23[['Ladder score','Social support','Healthy life expectancy','Freedom to make life choices','Generosity','Perceptions of corruption']].values[0],
    '2024': usa_values24[['Ladder score','social support','healthy life expectancy','freedom to make life choices','generosity','perceptions of corruption']].values[0]
}, index=['Ladder score','Social support','Healthy life expectancy','Freedom to make life choices','Generosity','Perceptions of corruption'])

print(comparison_df)

comparison_df['Change (%)'] = (comparison_df['2024'] - comparison_df['2023']) / comparison_df['2023'] * 100

print("\nPercentage Change from 2023 to 2024:")
comparison_df
```

The data indicates that while social support and generosity have improved significantly, the USA's happiness ranking is negatively impacted by setbacks in healthy life expectancy, freedom to make life choices, and perceptions of corruption. Addressing these issues, particularly restoring trust in institutions and enhancing public health measures, could help improve the overall happiness score in the future.

```python
import pandas as pd
gini_data=pd.read_excel('gini2018.xlsx')
gini_data
```

```python
gini_data.rename(columns={'country_name':'Country'}, inplace=True)
```

```python
gini_mapping = gini_data[['Country', 'value']].drop_duplicates()
```

```python
whr = whr.merge(gini_mapping, on='Country', how='left')
```

```python
# Fill missing Gini values with regional mean
whr['value'] = whr.groupby('Region')['value'].transform(
   lambda x: x.fillna(x.mean())
)
```

```python
whr['value'].mean()
```

```python
top_50 = whr.sort_values(by='Ladder score', ascending=False).head(50)
```

```python
plt.figure(figsize=(10, 6))
ax=sns.histplot(top_50['value'].dropna(), kde=True, bins=5, color='skyblue')
plt.title('Distribution of Gini Coefficient in Top 50 Happiest Countries')
plt.xlabel('Gini Coefficient')
plt.ylabel('Frequency')
for container in ax.containers:
    ax.bar_label(container)

```

A good number of countries (13) also fall in the range of 25-30, which shows even greater income equality.
This could indicate that income equality might be a contributing factor to happiness in these countries
Fewer countries (around 14 total) fall in the ranges 40-55, indicating that high income inequality is less common among the top happiest countries.

```python
top_50_countries = whr.nlargest(50, 'Ladder score')
correlations = top_50_countries[['value', 'social support','freedom to make life choices',
                         'healthy life expectancy','generosity']].corr()
print(correlations['value'])
```

Social support: The correlation of -0.401935 indicates a moderate negative correlation with the Gini coefficient. This means that as social support increases, the Gini coefficient tends to decrease (i.e., less income inequality). It suggests that stronger social support systems may help reduce inequality.

does climate risk vulnerabity impact happiness ?

```python
import pandas as pd
cri=pd.read_csv('climate-risk-index-1.csv')
cri.head()
```

```python
cri.rename(columns={'country':'Country'}, inplace=True)
```

```python
cri_mapping = cri[['Country', 'cri_rank','cri_score']].drop_duplicates()
```

```python
whr = whr.merge(cri_mapping, on='Country', how='left')
```

```python
whr['cri_rank'].fillna(whr['cri_rank'].mean(), inplace=True)
```

```python
whr['cri_score'].fillna(whr['cri_score'].mean(), inplace=True)
```

```python
whr.describe()
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(whr['cri_score'], kde=True, bins=20, color='skyblue')
plt.title('Distribution of CRI Scores')
plt.xlabel('CRI Score')
plt.ylabel('Frequency')
plt.show()


```

```python
# Correlation between CRI and Happiness
correlation_matrix = whr[['cri_score', 'cri_rank', 'Ladder score']].corr()
print(correlation_matrix)

# Visualize the correlation
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation: CRI vs Happiness")

```

there is no significant correlation between climate vulnerability (CRI Score and Rank) and happiness (Ladder score). This indicates that countries with high climate vulnerability do not necessarily report lower happiness levels, and vice versa.

The World Happiness Report emphasizes social, economic, and governance factors (like social support, GDP per capita, freedom to make choices, etc.) that may outweigh the direct effects of climate vulnerability on happines

While climate vulnerability does not appear to directly influence happiness scores at the national level, the broader effects of climate change on economies, health, and well-being may still pose challenges that are not fully captured in this dataset. Policymakers should continue to address climate challenges while fostering social and economic well-being to ensure a comprehensive approach to happiness and resilience.
